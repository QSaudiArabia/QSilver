<h3>Task 1 (on paper)</h3>
<a id="task1"></a>

Show that $\ket{-}$ and $\ket{+}$ are eigenvectors for the $X$ operator with eigenvalues -1 and 1 respectively.

<h3>Solution </h3>



\begin{align*}
X \ket{-} = X \frac {\ket{0} - \ket{1}} {\sqrt{2}}
& = \frac {\ket{1} - \ket{0}} {\sqrt{2}}\\
& = -\frac {\ket{0} - \ket{1}} {\sqrt{2}}\\
& = -\ket{-}
\end{align*}


\begin{align*}
X \ket{+} = X \frac {\ket{0} + \ket{1}} {\sqrt{2}}
& = \frac {\ket{1} + \ket{0}} {\sqrt{2}}\\
& = \frac {\ket{0} + \ket{1}} {\sqrt{2}}\\
& = \ket{+}
\end{align*}

Hence $ \ket{-}$ and $\ket{+}$ are eigenvectors of $ X $ operator with the eigenvalues -1 and 1 respectively.

<h3>Task 2 (on paper)</h3>
<a id="task2"></a>

Consider the following quantum state where $ x=0 $ or $ 1 $. How can you find out the value of $x$ by applying a single operator?

$$ 
\frac {\ket{0} + (-1)^x \ket{1}} {\sqrt{2}}
$$


<h3>Solution </h3>

If our aim is to find out $ x $, then applying a Hadamard gate we can determine whether $ x=0 $ or $ 1 $ depending on the outcome. If the outcome is $ \ket{0} $, then $ x=0 $ and if the outcome is $ \ket{1} $, then $ x=1 $.

<h3>Task 3</h3>

Consider the unitary operator $U$ with eigenvector $\ket{1}$ and eigenvalue $e^{2\pi i \phi}$ where $\phi = \frac{5}{16}$. $CU$ operator can be realized as follows:

```python
from qiskit.circuit.library import CPhaseGate
from math import pi

phase = 5/16
# e^{i*2*pi*phase} on |11>: Qiskit's CPhaseGate(theta) applies e^{i*theta}, so theta = 2*pi*phase
CU = CPhaseGate(2 * pi * phase)
```

Pick $t=4$ and implement the phase estimation circuit to estimate $\phi$. Write a function named <i>qpe</i> which takes as parameters <i>t</i>, <i>control</i> qubits, <i>target</i> qubits, <i>circuit</i> and the operator <i>CU</i>.

Note that you will get an exact result since $t=4$ is precise enough.

<i>Remarks:</i>

- You can apply arithmetical operations to gates, such as taking powers — every Qiskit `Gate` supports `.power(k)`.
- To apply $QFT^{\dagger}$, you can use the function you have written in the previous notebook (`iqft.py`).
- Some code is already given to guide you.


<h3>Solution </h3>

```python
# from iqft.py
from iqft import iqft
```

```python
#%%writefile qpe.py

def qpe(t, control, target, circuit, CU):

    #Apply Hadamard to control qubits
    for q in control:
        circuit.h(q)

    #Apply CU gates
    for i in range(t):
        #Obtain the power of CU gate
        CUi = CU.power(2**i)
        #Apply CUi gate where t-i-1 is the control
        circuit.append(CUi, [control[t - i - 1]] + list(target))

    #Apply inverse QFT
    iqft(t, control, circuit)

    return circuit
```

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator


def most_likely_estimate(counts, t):
    """Turn AerSimulator counts for the control register into (x, phi_estimate).

    Qiskit's bitstrings are printed with qubit 0 as the right-most
    character; we defined control[0] to be the *first*/most-significant
    control qubit throughout (same convention as iqft.py), so we reverse
    the substring before reading it as an integer.
    """
    best_bits = max(counts, key=counts.get)
    x = int(best_bits[-t:][::-1], 2)
    return x, x / 2**t


#Create circuit
t = 4  #Number of qubits in the control register
n = 1  #Number of qubits in the register storing eigenvector

circuit = QuantumCircuit(t + n, t)

#Create control and target qubits (just use qubit indices 0..t+n-1)
control = list(range(t))
target = list(range(t, t + n))

#Set target qubit to state |1>
circuit.x(target)

#Apply QPE
qpe(t, control, target, circuit, CU)

#Measure and print the results
circuit.measure(control, range(t))

sim = AerSimulator()
print('Sample the circuit:')
result = sim.run(transpile(circuit, sim), shots=1000).result()
counts = result.get_counts()
print(counts)
x, est = most_likely_estimate(counts, t)
print('most likely x:', x, ' estimate for phi:', est)
print(circuit)
```

<h3>Task 5</h3>

Try Task 3 this time by setting $\phi=0.685$. Calculate and print the estimates for different $t$ values between 1 and 10. 

<h3>Solution </h3>

```python
phase = 0.685
CU = CPhaseGate(2 * pi * phase)
```

```python
for t in range(1, 10):

    n = 1
    circuit = QuantumCircuit(t + n, t)
    control = list(range(t))
    target = list(range(t, t + n))

    #Set target qubit to state |1>
    circuit.x(target)

    #Apply QPE
    qpe(t, control, target, circuit, CU)

    circuit.measure(control, range(t))

    sim = AerSimulator()
    result = sim.run(transpile(circuit, sim), shots=1000).result()
    counts = result.get_counts()

    #Most frequent observation
    x, est = most_likely_estimate(counts, t)
    print("t:", t, "estimation:", est)
```
