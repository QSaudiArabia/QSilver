<h3>Task 1</h3>

Let $N=111$. What percentage of the elements that are less than $N$ are relatively prime with $N$? Write a Python code to find out. (You can use gcd function of numpy.)

<h3>Solution</h3>

```python
import numpy as np

#Create an empty list
rlist = []
N = 111

#If relatively prime with N, append to the list
for i in range(1, N):
    if np.gcd(N, i) == 1:
        rlist.append(i)

print(rlist)

print(len(rlist) * 100 / N, "percentage of the integers are relatively prime with", N)
```

<a id="task2"></a>
<h3>Task 2</h3>

Calculate the order of each element $ x $ that is relatively prime with $N $. What percentage of the $ x $’s have even order and satisfy $x^{r/2} \neq -1 \Mod{N}$?

Put the elements that satisfy the conditions in a dictionary together with their order.


<h3>Solution</h3>

```python
import numpy as np
counter = 0
#This will hold the list of integers that satisfy the conditions together with the order
satisfy = {}

#rlist contains the relatively prime numbers with N
for i in range(len(rlist)):
    r = 1
    while True:
        if (rlist[i]**r) % N == 1:
            if r % 2 == 0 and ((rlist[i]**int(r / 2)) % N != N - 1):
                counter = counter + 1
                print("Order of", rlist[i], ":", r)
                satisfy[rlist[i]] = r
            break
        r = r + 1

print(counter * 100 / N, "percentage of the integers satisfy the conditions")
```

<a id="task3"></a>
<h3>Task 3</h3>

Pick randomly one of the $x$ you found in Task 2 and calculate gcd$(x^{r/2}-1,N)$ and gcd$(x^{r/2}+1,N)$.

<h3>Solution</h3>

```python
import random
import math  # math.gcd handles arbitrarily large ints; np.gcd overflows
             # here since x**(r/2) can exceed 64 bits for some (x, r) pairs

#Pick a random x and its order from the dictionary we have created above
x, r = random.choice(list(satisfy.items()))
print(x, "is picked with order", r)

#Calculate gcd
print("Factors of", N, ":", math.gcd((x**int(r / 2) - 1), N), "and", math.gcd((x**int(r / 2) + 1), N))
```

<a id="task4"></a>
<h3>Task 4</h3>

Factor 21 using Shor's Algorithm.

- Pick a random $x$ which is relatively prime with 21.
- Apply phase estimation circuit to the operator $U_x$.
- Use continued fractions algorithm to find out $r$.
- Compute $gcd(x^{r/2} -1, N)$ and $gcd(x^{r/2}+1, N)$

<h3>Solution</h3>

```python
N = 21
```

```python
#Pick a random x relatively prime with N
import random as rand
import numpy as np

counter = 0

while True:
    x = rand.randrange(1, N)
    counter = counter + 1
    if np.gcd(x, N) == 1:
        break

print(x, " is picked after ", counter, " trials")
```

```python
#Load the function Ux
from operator_cu import Ux
```

```python
#Create CU operator by calling function Ux(x,N)
CU = Ux(x, N)
```

```python
#Load the phase estimation function
from qpe import qpe
```

```python
#Load the inverse QFT function (qpe.py already uses it internally)
from iqft import iqft
```

```python
#Determine t and n, size of the control and target registers
import math

n = math.ceil(math.log2(N))
t = 2 * n + 1 + math.ceil(math.log2(2 + 1 / (2 * 0.1)))
print("n =", n, " t =", t)
```

```python
from qiskit import QuantumCircuit

#Create control and target qubits and the circuit
circuit = QuantumCircuit(t + n, t)

control = list(range(t))
target = list(range(t, t + n))
circuit.x(target[-1])

#Call phase estimation circuit with parameter CU
qpe(t, control, target, circuit, CU)

#Measure the control register
circuit.measure(control, range(t))
```

```python
#Run the circuit once (a single shot, matching a single run of the real algorithm)
from qiskit_aer import AerSimulator
from qiskit import transpile

sim = AerSimulator()
result = sim.run(transpile(circuit, sim), shots=1).result()
counts = result.get_counts()
print(counts)
```

```python
#Convert the measured bitstring to an integer.
#Qiskit prints bitstrings with qubit 0 (=the *last* control qubit
#appended, per qpe.py) on the right, but we want control[0] to be the
#most-significant bit of the result (matching iqft.py's convention) -
#so reverse before converting.
bits = list(counts.keys())[0]
b = int(bits[-t:][::-1], 2)
print(b)
```

```python
#Load the contFrac and convergents functions
import sys
sys.path.append('../include')
from helpers import contFrac, convergents

#Run continued fractions algorithm to find out r
cf = contFrac(b / (2**t))
print(cf)
cv = convergents(cf)
print(cv)
```

The candidates for $r$ are the denominators of the convergents above.

```python
#Check if r is even, and x^(r/2) is not equal to -1 Mod N, and that it's
#a genuine order of x (x**r % N == 1) - try each convergent's denominator,
#largest first, and re-run the circuit for a fresh sample if none work
#(exactly the "you may not get r on the first trial, repeat" case below).

def try_recover_r(x, N, cv):
    for frac in sorted(cv, key=lambda f: -f.denominator):
        r = frac.denominator
        if r > 0 and pow(x, r, N) == 1 and r % 2 == 0 and pow(x, r // 2, N) != N - 1:
            return r
    return None


r = try_recover_r(x, N, cv)
print("r =", r)
```

Note that you may not be able to get the $r$ value in your first trial. In such a case, you need to repeat the algorithm. The cell below does that automatically: if the single shot above didn't yield a usable $r$, it keeps sampling fresh shots (each one a fresh run of the phase estimation circuit) until one does. There's a subtlety worth knowing: Shor's theorem only guarantees this succeeds for *at least* half of the valid choices of $x$ (Theorem 2 below) - for the rest, $x^{r/2}$ always equals $-1 \Mod{N}$ for the true order $r$, no matter how many shots you sample, so resampling forever would never finish. The cell below handles this too: after a number of unlucky shots with the same $x$, it picks a fresh $x$ (and rebuilds the circuit for it) rather than resampling the same one indefinitely.

```python
#A "bad" x (x^(r/2) == -1 Mod N for its true order) can never succeed no
#matter how many shots - resample a handful of times with the current x,
#then pick a fresh one and rebuild the circuit if it's still not working.
max_attempts_per_x = 20
attempts = 1
attempts_with_this_x = 0

while r is None:
    attempts_with_this_x += 1
    if attempts_with_this_x > max_attempts_per_x:
        while True:
            x = rand.randrange(1, N)
            if np.gcd(x, N) == 1:
                break
        print("This x looked unlucky - trying a new x =", x)
        CU = Ux(x, N)
        circuit = QuantumCircuit(t + n, t)
        circuit.x(target[-1])
        qpe(t, control, target, circuit, CU)
        circuit.measure(control, range(t))
        attempts_with_this_x = 0

    result = sim.run(transpile(circuit, sim), shots=1).result()
    counts = result.get_counts()
    bits = list(counts.keys())[0]
    b = int(bits[-t:][::-1], 2)
    cf = contFrac(b / (2**t))
    cv = convergents(cf)
    r = try_recover_r(x, N, cv)
    attempts += 1

print("Found r =", r, "after", attempts, "attempt(s), with x =", x)
```

Now let's check $gcd(x^{r/2} -1, N)$ and $gcd(x^{r/2}+1, N)$.

```python
#Compute gcd to find out the factors of N
#(math.gcd, not np.gcd - x**(r/2) can be astronomically large if a
#spuriously-passing large r ever comes out of try_recover_r, and np.gcd
#overflows its C-long internals on a Python int that big; math.gcd doesn't.)
import math
print("Factors of", N, ":", math.gcd((x**int(r / 2) - 1), N), "and", math.gcd((x**int(r / 2) + 1), N))
```

Whatever $x$ was drawn above, this recovers the factors of 21 (3 and 7) the same way Shor's algorithm always does: quantum phase estimation for the order $r$ of $x \Mod{21}$, then classical continued fractions and gcd to turn that $r$ into the factors. (For example, with $x=10$ — what the original Cirq notebook's own random draw happened to pick — this recovers $r=6$, matching $10$'s classical order mod 21, and gcd gives exactly 3 and 7.)
